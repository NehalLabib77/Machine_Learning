# Pandas Data Manipulation - Beginner to Intermediate Guide

## 🎯 What is Pandas?
Pandas is Python's primary library for working with data. If your data was a spreadsheet (like Excel), Pandas lets you manipulate it programmatically. It's the bridge between raw data and data analysis!

## 📊 Why Learn Pandas?
- **Handle Real Data**: Load and work with messy, real-world data
- **Quick Analysis**: Calculate statistics without complex code
- **Data Cleaning**: Fix missing values, remove duplicates, format data
- **Integration**: Perfect tool before visualizing or modeling your data
- **Industry Standard**: Used by most data professionals

## 🗂️ Core Data Structures

### Series (1D Data)
- Like a column in a spreadsheet
- One dimension with index labels
- Example: list of student grades

### DataFrame (2D Data)
- Like a full spreadsheet with rows and columns
- Multiple dimensions (rows and columns)
- Example: employee data with name, salary, department

## 📚 Beginner's Learning Path

### Section 1: Core Concepts & Getting Started
**Goal**: Understand basic data structures and how to create them

**Topics Covered**:
- Import Pandas and NumPy
- Create a Series from a list
- Create a Series with custom index labels
- Create a DataFrame from a dictionary
- Understand DataFrame structure

**Why This Matters**: You need to know how to create and understand data before you can analyze it

### Section 2: Exploring Your Data (The First Steps)
**Goal**: Learn how to look at and understand your data

**Key Methods**:
- `df.info()` - Get overview of data (types, missing values)
- `df.head()` - See first few rows
- `df.tail()` - See last few rows
- `df.shape` - How many rows and columns?
- `df.columns` - What are the column names?
- `df.describe()` - Statistical summary (mean, min, max, etc.)
- `df.dtypes` - What data types do we have?

**Why This Matters**: Before analyzing data, you MUST understand it first

### Section 3: Selecting Data (Finding What You Need)
**Goal**: Learn different ways to extract data from your DataFrame

**Techniques**:
- Select single column: `df['ColumnName']`
- Select multiple columns: `df[['Col1', 'Col2']]`
- Select rows by position: `df.iloc[0:5]` (first 5 rows)
- Select rows by label: `df.loc[0:2]` (rows labeled 0-2)
- Filter by condition: `df[df['Age'] > 30]` (all rows where Age > 30)

**Why This Matters**: You rarely want all data; you need to focus on specific parts

### Section 4: Data Manipulation (Making Changes)
**Goal**: Learn how to add, remove, and modify data

**Operations**:
- Add new column: `df['NewCol'] = values`
- Remove column: `df.drop('ColumnName', axis=1)`
- Rename columns: `df.rename(columns={'OldName': 'NewName'})`
- Sort data: `df.sort_values('ColumnName')`
- Apply functions: `df['Column'].apply(lambda x: x * 2)`

**Why This Matters**: Real data needs cleaning and organizing

### Section 5: Handling Missing Data (Dealing with Problems)
**Goal**: Find and fix incomplete data

**Issues & Solutions**:
- Identify missing: `df.isnull().sum()` - How much missing data?
- Fill missing: `df.fillna(0)` - Replace with specific value
- Drop missing: `df.dropna()` - Remove rows with missing values
- Forward fill: `df.fillna(method='ffill')` - Use previous value

**Why This Matters**: Most real-world data has gaps that you must handle

### Section 6: Data Aggregation (Summarizing Data)
**Goal**: Calculate summary statistics and group data

**Operations**:
- Mean: `df['Column'].mean()` - Average value
- Sum: `df['Column'].sum()` - Total
- Count: `df['Column'].count()` - How many non-missing?
- Group By: `df.groupby('Category')['Value'].mean()` - Average per category
- Pivot tables: Create summaries organized by multiple dimensions

**Why This Matters**: Turn raw data into actionable insights

### Section 7: Combining Data (Merging Multiple Sources)
**Goal**: Join data from different sources

**Techniques**:
- Concatenate: Combine rows from multiple DataFrames
- Merge: Join based on common columns (like SQL JOIN)
- Join: Combine based on index

**Why This Matters**: Real analysis often requires combining data from multiple sources

### Section 8: Loading & Saving Data (In & Out)
**Goal**: Work with data files

**File Formats**:
- CSV: `pd.read_csv('file.csv')` - Most common format
- Excel: `pd.read_excel('file.xlsx')` - Spreadsheet format
- JSON: `pd.read_json('file.json')` - Web data format
- Database: SQL queries directly to DataFrame

**Why This Matters**: Data lives in files; you need to load and save it

## 🎯 Step-by-Step Example
```python
# 1. Load data
df = pd.read_csv('data.csv')

# 2. Explore data
print(df.head())          # First look
print(df.info())          # Summary
print(df.describe())      # Statistics

# 3. Clean data
df = df.dropna()          # Remove missing values
df = df.drop_duplicates() # Remove duplicates

# 4. Analyze data
average = df['Age'].mean() # Calculate average
by_group = df.groupby('Department')['Salary'].mean() # Group analysis

# 5. Visualize
df.plot(kind='bar')       # Quick visualization
```

## 💡 Common Questions Beginners Ask

**Q: What's the difference between Series and DataFrame?**
A: Series is 1 column; DataFrame is multiple columns (like full spreadsheet)

**Q: Why use Pandas instead of Excel?**
A: For large datasets, complex operations, and reproducible analysis

**Q: How do I handle missing data?**
A: Decide: Drop it, Fill it, or Leave it based on your analysis

**Q: What's groupby?**
A: Split data into groups, apply function, combine results. Like PIVOT in Excel.

#### Data aggregating and grouping 

````python
df = pd.read_csv('coffee.csv')
````

````python
df.head()
````

````python
grp_mean = df.groupby('Coffee Type')['Units Sold'].mean()
print(grp_mean)
````

````python
grp_sum = df.groupby(['Day','Coffee Type'])['Units Sold'].sum()
print(grp_sum)
````

````python
# Multiple aggregate function 

grp_agg = df.groupby(['Day'])['Units Sold'].agg(['mean','sum','count'])
print(grp_agg)
````

#### Merging and joining dataframes 

````python
df1 = pd.DataFrame({'Key':['A','B','C'],'Value1':[1,2,3]})
df2 = pd.DataFrame({'Key':['A','B','D'],'Value1':[4,5,6]})
````

````python
df1
````

````python
df2
````

````python
# Returns only the rows that match in BOTH tables.
pd.merge(df1,df2, on ='Key', how = 'inner')
````

````python
# Returns matching rows PLUS non-matching rows (with NULL values).
pd.merge(df1,df2, on ='Key', how = 'outer')
````

````python
pd.merge(df1,df2, on ='Key', how = 'left')
````

````python
pd.merge(df1,df2, on ='Key', how = 'right')
````
