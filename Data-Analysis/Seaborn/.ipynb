**BOXPLOT**

A boxplot (a.k.a. whisker plot) is a chart that shows the spread of data (how values are distributed).

**It summarizes your data in 5 key numbers:**

Minimum → the smallest value (ignoring outliers).

Q1 (25th percentile) → 25% of values are below this.

Median (50th percentile) → the middle value.

Q3 (75th percentile) → 75% of values are below this.

Maximum → the largest value (ignoring outliers).

Outliers (very high/low unusual values) are usually shown as dots.

**How to read it**

The box = middle 50% of the data (between Q1 and Q3).

The line inside box = median (the middle).

The whiskers (lines outside the box) = range of most values.

The dots = outliers (unusual points).

```python
sns.boxplot(tips['tip'])
```

```python
sns.catplot(data = tips , x = 'day', y = 'total_bill', kind = 'box')
```

```python
sns.catplot(data = tips , x = 'day', y = 'total_bill', kind = 'box', hue = 'sex')
```

Boxplot = just summary (median + spread + outliers).

Violin plot = summary plus distribution shape.

**VIALIN-PLOT**

What is a Violin Plot?

A violin plot shows the distribution (shape) of data for a variable, kind of like a fatter, smoother boxplot.

It combines a boxplot (median + quartiles)

with a KDE (Kernel Density Estimate) — which shows how the data is spread (the “violin” shape).


**How to read it**

The thick white dot = median.

The thick box inside = interquartile range (Q1 → Q3).

The thin line (“whisker”) = range of data.

The shape around it = distribution curve. Wider parts = more data points there; thinner parts = fewer.

```python
sns.violinplot(tips['total_bill'])
```

```python
sns.catplot(data = tips , x = 'day', y = 'total_bill', kind = 'violin')
```

```python
sns.catplot(data = tips , x = 'day', y = 'total_bill', kind = 'violin', hue = 'sex')
```

```python
sns.catplot(data = tips , x = 'day', y = 'total_bill', kind = 'violin', hue = 'sex', split = True)
```

**BARPLOT**

Purpose: Shows the average (or another estimator) of a numerical variable for each category.

Needs: one categorical variable (x) + one numerical variable (y).

Default behavior: calculates the mean of the numerical values in each category.

Can also show confidence intervals (error bars).

```python
sns.catplot(data = tips , x = 'smoker', y = 'total_bill', kind = 'bar')
```

```python
sns.catplot(data = tips , x = 'smoker', y = 'total_bill', estimator = np.std, kind = 'bar', hue = 'sex') 

# Short error bars → less variability, more confidence.
# Long error bars → lots of variability, less confidence.
```

**COUNTPLOT**

Purpose: Shows the count (frequency) of each category.

Needs: only a categorical variable (x or y).

It’s basically a histogram for categorical data.

No numerical variable needed!

```python
sns.catplot(data = tips , x = 'sex', kind = 'count', hue = 'smoker')
```

**HEATMAPS**

A heatmap = numbers shown as colors → quick way to see patterns, highs/lows, and relationships.

A heatmap is just a table of numbers that’s turned into colors so you can quickly see patterns.

Big values = darker/stronger color

Small values = lighter/softer color

It’s like a colored Excel sheet that makes it easier to spot trends.

```python
flights = sns.load_dataset('flights')
flights
```

```python
x = flights.pivot_table(index = 'year', columns = 'month', values = 'passengers', aggfunc = 'sum')
```

```python
sns.heatmap(x)
```

```python
sns.heatmap(x, linewidth = .5)
```

```python
sns.heatmap(x, cbar = False , linewidth = .5, annot = True , fmt = 'd')
```

```python
sns.heatmap(x, linewidth = .5, annot = True , fmt = 'd', cmap = 'summer')
```

```python
sns.heatmap(x,linewidth = .5, annot = True , fmt = 'd', cmap = 'viridis')
```

**CLUSTERMAP**

A clustermap is like a heatmap + clustering (grouping similar rows/columns together).

It shows your data as a colored matrix (like a heatmap).

BUT it also rearranges rows and columns so that similar values are grouped side by side.

It adds dendrograms (tree-like diagrams) on top and left to show the grouping.

```python
sns.clustermap(x)
```

```python
sns.clustermap(x, metric = 'correlation')
```

```python
sns.clustermap(x, z_score = 0, metric = 'correlation')
```

```python
sns.clustermap(x, z_score = 0,annot = True,  metric = 'correlation')
```

**JOINTPLOT**

A jointplot = scatterplot + histograms → lets you see both the relationship between two variables and their individual distributions.

**The kind parameter changes the center plot:**

"scatter" → dots (default).

"hex" → hexbin plot (good for large data).

"kde" → smooth density plot.

"reg" → scatter + regression line.


```python
tips
```

```python
sns.jointplot(data = tips , x = 'total_bill', y = 'tip')
```

```python
sns.jointplot(data = tips , x = 'total_bill', y = 'tip', kind = 'hex')
```

```python
sns.jointplot(data = tips , x = 'total_bill', y = 'tip', kind = 'kde')
```

```python
sns.jointplot(data = tips , x = 'total_bill', y = 'tip', kind = 'reg')
```

**PAIRPLOT**

A pairplot in Seaborn is a grid of plots that shows the relationship between every pair of variables in your dataset.

It makes a scatterplot for each pair of numerical variables.

It puts a histogram (or KDE) on the diagonal to show each variable’s distribution.

So, it’s like a “full picture” of how all the variables relate to each other.

```python
iris = sns.load_dataset('iris')
iris
```

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris, hue = 'species')
```
