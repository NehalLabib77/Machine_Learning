**IMPORTING LIBRARIES**

```python
# Import required visualization and data handling libraries
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# Check Seaborn version
print(sns.__version__)
```

```python
# Load Iris dataset (classic dataset for visualization examples)
iris = sns.load_dataset('iris')
iris.head()
```

```python
# Create a scatter plot showing relationship between two variables
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width')
plt.show()
```

```python
# Create a line plot to show trends over a continuous variable
sns.lineplot(data=iris, x='sepal_length', y='sepal_width')
```

### Bar plot

```python
tips = pd.read_csv('tips.csv')
tips.head()
```

```python
sns.barplot(data = tips, x = 'day', y = 'total_bill')
```

```python
# Create a histogram to show distribution of a continuous variable
sns.histplot(tips['total_bill'], bins = 10, kde=True)
```

```python
# Create a KDE (Kernel Density Estimate) plot showing smooth distribution
sns.kdeplot(tips['total_bill'], fill = True)
```

```python
# Create a box plot to show distribution and outliers by category
sns.boxplot(data=tips, x='day', y='total_bill')
```

**Categorical Scatterplots**

```python
# Load tips dataset for additional examples
tips = pd.read_csv('tips.csv')
tips.head()
```

```python
# Create a bar plot showing average values by category
sns.barplot(data=tips, x='size', y='total_bill', hue='sex')
```

```python
# Create a violin plot combining box plot and KDE information
sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
# Create a strip plot showing individual data points
sns.stripplot(data=tips, x='sepal_length', y='sepal_width')
```

```python
# Create a count plot showing frequency of categories
sns.countplot(data=tips, x='sepal_length')
```

### pair plot 
# Most easy way to understand relationship

```python
sns.pairplot(data = tips)
```

### Correlation 

```python
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
# Create a regression plot showing linear relationship with confidence interval
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
# Create a lmplot (regression plot) with faceting by category
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
# Create a lmplot with faceting by multiple dimensions
sns.lmplot(data=tips, x='total_bill', y='tip', col='time', row='sex')
```

```python
# Create a categorical plot with more flexibility (similar to lmplot)
sns.catplot(data=tips, x='day', y='total_bill', hue='sex', kind='bar')
```

```python
# Create a relational plot with interactive exploration of relationships
sns.relplot(data=tips, x='total_bill', y='tip', hue='sex', size='party_size')
```

```python
# Create a multiple subplot grid showing different variables
g = sns.relplot(data=tips, x='total_bill', y='tip', col='time', row='sex', hue='day')
g.fig.suptitle('Tips Analysis by Time and Gender', y=1.02)
```

```python
# Create a clustermap showing hierarchical clustering of data
iris_numeric = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
sns.clustermap(iris_numeric.sample(20))
```

**HEATMAPS**

A heatmap = numbers shown as colors → quick way to see patterns, highs/lows, and relationships.

A heatmap is just a table of numbers that’s turned into colors so you can quickly see patterns.

Big values = darker/stronger color

Small values = lighter/softer color

It’s like a colored Excel sheet that makes it easier to spot trends.

```python
# Calculate correlation matrix between numerical variables
corr = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()
corr
```

```python
# Create a heatmap visualization of correlation matrix
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
```

```python
# Create a heatmap with additional styling and cbar
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', vmin=-1, vmax=1)
```

```python
# Create a pairplot showing pairwise relationships between all variables
sns.pairplot(iris)
```

```python
# Create a pairplot with hue coloring by species category
sns.pairplot(iris, hue='species')
```

```python
# Create a pairplot with different scatter plot kinds
sns.pairplot(iris, diag_kind='kde', plot_kws=dict(s=50, alpha=0.5))
```

**CLUSTERMAP**

A clustermap is like a heatmap + clustering (grouping similar rows/columns together).

It shows your data as a colored matrix (like a heatmap).

BUT it also rearranges rows and columns so that similar values are grouped side by side.

It adds dendrograms (tree-like diagrams) on top and left to show the grouping.

```python
# Create a joyplot-like visualization using FacetGrid
g = sns.FacetGrid(tips, col='sex', height=4)
g.map(sns.histplot, 'total_bill')
```

```python
# Create faceted scatter plot with custom styling
g = sns.FacetGrid(tips, col='day', hue='sex', height=4)
g.map(sns.scatterplot, 'total_bill', 'tip')
```

```python
# Create a styled plot with different style context
sns.set_style('darkgrid')
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
```

```python
# Create a plot with white background style
sns.set_style('white')
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex')
```

```python
# Create a joint plot showing marginal distributions
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

**PAIRPLOT**

A pairplot in Seaborn is a grid of plots that shows the relationship between every pair of variables in your dataset.

It makes a scatterplot for each pair of numerical variables.

It puts a histogram (or KDE) on the diagonal to show each variable’s distribution.

So, it’s like a “full picture” of how all the variables relate to each other.

```python
iris = sns.load_dataset('iris')
iris
```

```python
# Create a joint plot with KDE for smooth density representation
sns.jointplot(data=tips, x='total_bill', y='tip', kind='kde')
```
