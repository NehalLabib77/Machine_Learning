## Data encoding is the process of converting data from one format to another so that it can be used effectively by computers or ML models.

In ML, encoding usually refers to transforming categorical or text data into numerical form, because most models only understand numbers.

### One-Hot Encoding
Converts each category(Column) into a binary vector.

Each category becomes a column with 0 or 1.

Color = ["Red", "Green", "Blue"]

One-Hot Encoded →

Red   → [1, 0, 0]

Green → [0, 1, 0]

Blue  → [0, 0, 1]


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'Color': ['Red', 'Green', 'Blue', 'Red', 'Green']})
df.head()
```

```python
# Create an instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[['Color']]).toarray()
```

```python
import pandas as pd 
encoder_df = pd.DataFrame(encoded,columns= encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['Blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis = 1)
```

## Label Encoding
Label Encoding is a data preprocessing technique used in Machine Learning to convert categorical (text) data into numerical form, because ML algorithms work with numbers, not strings.

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['Color']])
```

```python
lbl_encoder.transform([['Red']])
```

### Ordinal - Encoding
When a categorical feature has a meaningful order, ordinal encoding preserves that order.

High School → 0

Bachelor    → 1

Master      → 2

PhD         → 3

```python
from sklearn.preprocessing import OrdinalEncoder
df = pd.DataFrame({'data' : ['Low', 'Medium', 'High', 'High','Low']})
df
```

```python
encoder = OrdinalEncoder(categories = [['Low','Medium','High']])
```

```python
encoder.fit_transform(df[['data']])
```

### Target-Guided Ordinal Encoding
(also called Target Encoding with ordering) is an advanced categorical encoding technique where categories are ordered and encoded based on their relationship with the target variable.

```python
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320]
})
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict() # Group by city ...mean on price 
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price) # .map() replaces values in a column using a mapping
df
```
