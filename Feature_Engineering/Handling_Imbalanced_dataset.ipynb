```python
# 🔹 Without a fixed seed → every run gives different random values

# 🔹 With the same seed → every run gives the same sequence of random values
import random

random.seed(20)
print(random.random())
print(random.random())
```

```python
import pandas as pd 
import numpy as np
n_sample = 1000
class_ratio_0 = .9
n_class_0 = int(n_sample * class_ratio_0)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0,n_class_1
```

## CREATE MY DATAFRAME WITH IMBALANCED DATASET

```python
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0,class_1]).reset_index(drop = True)
```

#### It creates n_class_0 random numbers drawn from a normal distribution with mean = 0 and standard deviation = 1, and assigns them to a feature called feature_1

```python
df.head()
```

```python
df['target'].value_counts()
```

### UP_sampling

```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

Resampling for imbalanced data (ML)

Goal: help model learn minority class

How it works:

Oversampling: duplicate or synthesize minority samples

Undersampling: remove majority samples

```python
from sklearn.utils import resample

df_minority_upsampled = resample(df_minority, replace = True,
         n_samples = len(df_majority),
         random_state = 42 # fixed seed
           )
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

## Down_Sampling --Bad for losing data

```python
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0,class_1]).reset_index(drop = True)
```

```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample

df_majority_downsampled = resample(df_majority, replace = False,
         n_samples = len(df_minority),
         random_state = 42 # fixed seed
           )
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority, df_majority_downsampled])
```

```python
df_downsampled['target'].value_counts()
```
