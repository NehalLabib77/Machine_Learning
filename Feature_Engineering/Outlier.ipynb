### 5 number summary(Boxplot describes this things)
1. Minimum
2. Q1
2. Median
3. Q3
4. Maximum

```python
import numpy as np
```

```python
list_num = [7, 23, 4, 19, 42, 11, 35, 2, 28, 16,
9, 31, 14, 6, 25, 38, 1, 20, 33, 12]

minimum,Q1,median,Q3,maximum = np.quantile(list_num,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
# For outlier we find IQR - It measures the spread of the middle 50% of the data
IOR = Q3 - Q1
print(IOR)
```

```python
lower_fence = Q1 - 1.5*(IOR)
higher_fence = Q3 + 1.5*(IOR)
```

```python
lower_fence,higher_fence
```

```python
import seaborn as sns
sns.boxplot(list_num)
```
