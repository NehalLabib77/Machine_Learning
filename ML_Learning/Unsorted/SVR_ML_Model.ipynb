# Label Encoding one hot Encoding

```python
df.columns
```

```python
x = df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y = df['total_bill']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.25,random_state = 42)
```

```python
x_train.head()
```

👉 Train = fit_transform()
👉 Test / New = transform()

```python
# Label Encoder 

from sklearn.preprocessing import LabelEncoder
l1 = LabelEncoder()
l2 = LabelEncoder()
l3 = LabelEncoder()

x_train['sex'] = l1.fit_transform(x_train['sex'])
x_train['smoker'] = l2.fit_transform(x_train['smoker'])
x_train['time'] = l3.fit_transform(x_train['time'])
```

```python
x_train.head()
```

```python
x_test['sex'] = l1.transform(x_test['sex'])
x_test['smoker'] = l2.transform(x_test['smoker'])
x_test['time'] = l3.transform(x_test['time'])
```

```python
x_test.head()
```

```python
# OneHot encoding - ColumnTransformer 

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# “Take column 3 → convert categorical values into numeric using one-hot encoding (dropping first column), and keep all other columns as they are.
ct = ColumnTransformer(transformers=[('onehot', OneHotEncoder(drop = 'first'),[3])], remainder='passthrough')
```

```python
x_train = ct.fit_transform(x_train)
x_train
```

```python
x_test = ct.transform(x_test)
x_test
```

```python
from sklearn.svm import SVR
svr = SVR()
```

```python
svr.fit(x_train,y_train)
```

```python
y_pred = svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

# Hyper parameter tuning 

```python
from sklearn.model_selection import GridSearchCV
params = {
	"C": [100, 10, 1.0, 0.1, 0.01],
	"kernel": ["linear", "rbf", "poly", "sigmoid"],
	"gamma": ["scale", "auto"]
}

grid = GridSearchCV(SVR(), param_grid=params, refit=True, cv=5, verbose=3)
```

```python
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
grid_predict = grid.predict(x_test)
print(r2_score(y_test,grid_predict))
print(mean_absolute_error(y_test,grid_predict))
```
