```python
x = 10
y = 2
z = 11

if x >= y and x >= z:
    print("x is bigger")
elif y >= x and y >= z:
    print("y is bigger")
else:
    print("z is bigger")
```

**Question 5:** Write a Python program to calculate the factorial of a number.

```python
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

print(f"Factorial of {n} is {factorial}")
```
