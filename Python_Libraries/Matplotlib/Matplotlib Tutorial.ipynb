```python
# Import required libraries for plotting and data handling
import matplotlib.pyplot as plt
import numpy as np
```

#### Basic line plot

```python
# Define data for x and y axes
x = [0, 1, 2, 3, 4]
y = [0, 2, 4, 6, 8]

# Plot the data
plt.plot(x, y)

# Set the title with custom font
plt.title("Our First Graph!", fontdict={'fontname': 'Comic Sans MS', 'fontsize': 21})

# Label the x-axis
plt.xlabel("X Axis")

# Label the y-axis
plt.ylabel("Y Axis")

# Display the plot
plt.show()
```

#### Customized line plot

```python
x = [0, 1, 2, 3, 4]
y = [0, 2, 4, 6, 8]

plt.plot(x, y,
         label='2x',
         color='red',
         linewidth=2,
         marker='.',
         linestyle='--',
         markersize=15,
         markeredgecolor='black')
plt.grid(True)
```

#### MUltiplot 

```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [11,21,31,41,51]

plt.figure(figsize = (9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color = 'green')
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.plot(y1,x,color = 'red')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot(y1,x,color = 'blue')
plt.title("Plot 3")
```

### Bar plot 

```python
categories = ['A','B','C','D','E']
values = [4,6,3,8,5]

plt.bar(categories,values,color ='purple')
```

### Histogram

```python
data = [1,2,3,3,4,4,4,4,5,5,5,6,6,6,6,7,7,7,7,7]
plt.hist(data,bins=5,color = 'orange', edgecolor = 'black') # Bins = How many dividation
```

### Scatterplot

```python
A = [1,2,3,4,6,7]
B = [2,3,4,5,7,8]

plt.scatter(A,B,color="blue", marker = 'x')
```

### Pie chart

```python
labels = ['A','B','C','D','E']
sizes = [30,20,10,50,20]
colors = ['gold','red','pink','blue','green']
explode = (0.2,0,0,0,0) # Move out the 1st slice 

plt.pie(sizes,explode = explode, labels = labels, colors = colors, autopct = "%1.1f%%", shadow = True) 
```
